```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
# Load the cleaned dataset

df = pd.read_csv("../dataset/electricity_consumption_cleaned.csv")

print("Dataset Shape:", df.shape)
print("Columns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Separate features and target

target_column = "Appliances"

X = df.drop(columns=[target_column])
y = df[target_column]

print("Target column:", target_column)
print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())
```

```python
# Select numerical features only

X = X.select_dtypes(include=["number"])

print("Numerical features shape:", X.shape)
print("\nNumerical feature columns:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
from sklearn.ensemble import ExtraTreesRegressor

best_model = ExtraTreesRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

best_model.fit(X_train, y_train)

print("Extra Trees model trained successfully!")
```

```python
# Make predictions on test data

y_pred = best_model.predict(X_test)

print("Predictions completed successfully!")
print("\nFirst 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MODEL EVALUATION")
print("-----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

```python
evaluation_results = pd.DataFrame({
    "Metric": ["MAE", "RMSE", "R2 Score"],
    "Value": [mae, rmse, r2]
})

display(evaluation_results)
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values[:20],
    "Predicted": y_pred[:20]
})

display(comparison)
```

```python
print("FINAL MODEL SUMMARY")
print("===================")
print("Best Model  :", "Extra Trees Regressor")
print("MAE         :", round(mae, 4))
print("RMSE        :", round(rmse, 4))
print("R2 Score    :", round(r2, 4))
```
