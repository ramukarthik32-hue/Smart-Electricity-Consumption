```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../dataset/energydata_complete.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Missing values before cleaning:")

print(df.isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print("Data types before conversion:")

print(df.dtypes)
```

```python
df["date"] = pd.to_datetime(df["date"])

print("Date converted successfully!")

print(df["date"].dtype)
```

```python
print("Negative Appliances values:")

print((df["Appliances"] < 0).sum())

print("\nNegative lights values:")

print((df["lights"] < 0).sum())
```

```python
total_missing = df.isnull().sum().sum()

print("Total missing values after cleaning:", total_missing)
```

```python
print("========== CLEANED DATASET ==========")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("Duplicate rows:", df.duplicated().sum())

print(
    "Missing values:",
    df.isnull().sum().sum()
)

print(
    "Date type:",
    df["date"].dtype
)
```
