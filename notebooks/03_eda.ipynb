```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("../dataset/electricity_consumption_cleaned.csv")

print("Cleaned dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(df["Appliances"], bins=50)

plt.title("Distribution of Appliance Energy Consumption")
plt.xlabel("Appliances Energy Consumption (Wh)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(14, 5))

plt.plot(df["date"], df["Appliances"])

plt.title("Appliance Energy Consumption Over Time")
plt.xlabel("Date")
plt.ylabel("Appliances Energy Consumption (Wh)")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["lights"],
    df["Appliances"],
    alpha=0.4
)

plt.title("Lights vs Appliance Energy Consumption")
plt.xlabel("Lights Energy Consumption (Wh)")
plt.ylabel("Appliances Energy Consumption (Wh)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["T1"],
    df["Appliances"],
    alpha=0.4
)

plt.title("Temperature vs Appliance Energy Consumption")
plt.xlabel("Temperature T1 (°C)")
plt.ylabel("Appliances Energy Consumption (Wh)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["RH_1"],
    df["Appliances"],
    alpha=0.4
)

plt.title("Humidity vs Appliance Energy Consumption")
plt.xlabel("Humidity RH_1 (%)")
plt.ylabel("Appliances Energy Consumption (Wh)")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Feature Correlation Matrix")

plt.tight_layout()
plt.show()
```

```python
target_correlation = (
    correlation_matrix["Appliances"]
    .sort_values(ascending=False)
)

print("Correlation with Appliances:")
print(target_correlation)
```

```python
top_features = (
    target_correlation
    .abs()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 features related to Appliances:")
print(top_features)
```

```python
df["date"] = pd.to_datetime(df["date"])

df["hour"] = df["date"].dt.hour

print("Hour feature created successfully!")
```

```python
hourly_consumption = (
    df.groupby("hour")["Appliances"]
    .mean()
)

plt.figure(figsize=(12, 5))

plt.plot(
    hourly_consumption.index,
    hourly_consumption.values,
    marker="o"
)

plt.title("Average Appliance Consumption by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Average Consumption (Wh)")
plt.xticks(range(24))

plt.grid(True)
plt.show()
```

```python
print("========== EDA SUMMARY ==========")

print("Dataset shape:", df.shape)

print(
    "\nAverage Appliances consumption:",
    round(df["Appliances"].mean(), 2),
    "Wh"
)

print(
    "Minimum Appliances consumption:",
    df["Appliances"].min(),
    "Wh"
)

print(
    "Maximum Appliances consumption:",
    df["Appliances"].max(),
    "Wh"
)

print(
    "\nMost correlated features with Appliances:"
)

print(target_correlation.head(6))

print("\nEDA completed successfully!")
```
